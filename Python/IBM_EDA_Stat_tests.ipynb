# Phase 2: Exploratory Data Analysis (EDA)

```python
# Attrition Distribution    

attrition_rate = df["Attrition_bin"].mean() * 100

attrition_counts = df["Attrition"].value_counts()

print(f"\n── Attrition Distribution ──\n\nEmployees who left: {attrition_counts[1]} ({attrition_rate:.2f}%)")
print(f'Employees who stayed: {attrition_counts[0]}' f' ({100 - attrition_rate:.2f}%)')
print(f"\nAttrition Rate: {attrition_rate:.2f}%\n")
```

```python
# Attrition by Department

dept_attrition = df.groupby("Department")["Attrition"].agg(
    total_employees="count",
    left_employees=lambda x: (x == "Yes").sum(),
    attrition_rate=lambda x: (x == "Yes").mean() * 100
)
print("── Attrition Rate by Department ──\n")
dept_attrition.sort_values("attrition_rate", ascending=False).round(2)
```

```python
# Attrition by Job Role

job_attrition = df.groupby("JobRole")["Attrition"].agg(
    total_employees="count",
    left_employees=lambda x: (x == "Yes").sum(),
    attrition_rate=lambda x: (x == "Yes").mean() * 100
)
print("── Attrition Rate by Job Role ──\n")
job_attrition.sort_values("attrition_rate", ascending=False).round(2)
```

```python
# Salary Coparison
# Stayed VS Left VS Total
 
salary_attrition = df.groupby("Attrition")["MonthlyIncome"].agg(['min', 'max', 'mean', 'median', 'std']).round(2)
salary_attrition.index = ["Stayed", "Left"]
print("── Salary Coparison : Stayed vs Left vs Total ──\n")
salary_attrition.round(2)

# Add Total employee monthly analysis
total_salary_stats = df["MonthlyIncome"].agg(['min', 'max', 'mean', 'median', 'std']).round(2)
total_salary_stats = pd.DataFrame(total_salary_stats).T
total_salary_stats.index = ["Total"]

# Merge with attrition stats
salary_comparison = pd.concat([salary_attrition, total_salary_stats])
salary_comparison
```

```python
# Satisfaction & Performance Level Comparison
# # Stayed VS Left VS Total

metrics_columns = ["JobSatisfaction", "EnvironmentSatisfaction", "RelationshipSatisfaction","WorkLifeBalance", "PerformanceRating", "JobInvolvement",]
metrics_attrition = df.groupby("Attrition")[metrics_columns].mean().round(2)
metrics_attrition.index = ["Stayed", "Left"]
print("── Satisfaction & Performance Level Comparison : Stayed vs Left vs Total──\n")

metrics_attrition

total_metrics = df[metrics_columns].mean().round(2)
total_metrics = pd.DataFrame(total_metrics).T
total_metrics.index = ["Total"]
metrics_comparison = pd.concat([metrics_attrition, total_metrics])

metrics_comparison
```

```python
# Overtime Analysis
overtime_attrition = df.groupby("OverTime")["Attrition"].agg(
    total_employees="count",
    left_employees=lambda x: (x == "Yes").sum(),
    attrition_rate=lambda x: (x == "Yes").mean() * 100
)  

print("── Overtime Analysis ──\n")
overtime_attrition.round(2)
```

```python
# Correlation with Attrition Analysis

numeric_columns = df.select_dtypes(include=np.number)
numeric_columns = numeric_columns.corr()["Attrition_bin"].drop(["Attrition_bin", "EmployeeNumber"])

print("── Correlation with Attrition (Numeric Columns) ──\n")
numeric_columns.abs().sort_values(ascending=False).round(2)

coor_with_attrition = pd.DataFrame(numeric_columns.abs().sort_values(ascending=False).round(3))
coor_with_attrition
```

# Phase 3: STATISTICAL TESTING

Independent-Samples T-Test (numerical features) \
H₀: Mean of feature is the SAME for Stay vs Leave \
Reject H₀ if p < 0.05 → the feature differs significantly by attrition 

```python
# Independent-Samples T-Test (Stay vs Leave)

print("\n── Independent-Samples T-Test (numeric columns) (Stay vs Leave) ──\n")

ttest_results = []
stay_grp  = df[df["Attrition"] == "No"]
leave_grp = df[df["Attrition"] == "Yes"]
num_cols = df.select_dtypes(include=np.number).columns.drop(["Attrition_bin", "EmployeeNumber"])

for col in num_cols:
    t, p = stats.ttest_ind(stay_grp[col], leave_grp[col])
    ttest_results.append({"Feature": col,  "T-Statistic": t.round(3), "P-Value": p.round(4)})

ttest_df = pd.DataFrame(ttest_results).sort_values(["P-Value", "T-Statistic"], ascending=[True, False])
ttest_df

```

── Chi-Square Test (categorical features) ─── \
H₀: Feature is INDEPENDENT of Attrition \
Reject H₀ if p < 0.05 → significant association exists

```python
print("\n── 2.2 Chi-Square Test (Categorical vs Attrition) ──")


cat_cols= df.select_dtypes(include="object").columns.drop("Attrition")
chi2_results = []
for col in cat_cols:
    ct = pd.crosstab(df[col], df["Attrition"])
    chi2, p, dof, _ = stats.chi2_contingency(ct)
    chi2_results.append({"Feature": col, "Chi²": chi2.round(3), "P-Value": p.round(4), "dof": dof})

chi2_df = pd.DataFrame(chi2_results).sort_values(["P-Value", "Chi²"], ascending=[True, False])
chi2_df
```

── ANOVA — Do means differ across 3+ groups? ─── \
Test: Does MonthlyIncome differ across JobLevel / Department / JobRole? \
H₀: All group means are equal

```python
print("\n──  One-Way ANOVA : MonthlyIncome  vs  JobLevel , Department , JobRole ──")
anova_combos = [
    ("MonthlyIncome", "JobLevel"),
    ("MonthlyIncome", "Department"),
    ("MonthlyIncome", "JobRole")
]

Anova_result = []
for outcome, grouping in anova_combos:
    if outcome not in df.columns or grouping not in df.columns:
        continue
    groups = [g[outcome].dropna().values for _, g in df.groupby(grouping)]
    F, p   = stats.f_oneway(*groups)
    Anova_result.append({"Outcome": outcome, "Grouping": grouping, "F-Statistic": F.round(3), "P-Value": p.round(4)})
anova_df = pd.DataFrame(Anova_result).sort_values(["P-Value", "F-Statistic"], ascending=[True, False])
anova_df
```
